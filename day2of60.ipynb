# **Practice task **

Build a calculator that takes principal, rate and time for simple interest, and weight and height for BMI. Print the BMI category Underweight below 18.5, Normal 18.524.9, Overweight 2529.9, Obese 30 and above). Then take a full name and print it in title case, its initials, and the number of vowels.

```python
print("----------Simple calculator----------")
principal = float(input("Enter the principal amount:"))
rate = float(input("enter the rate:"))
time_y = float(input("Enter the time taken (year):"))

SI = (principal*rate*time_y)/100
print("the simple interest:", SI)
```

```python
print("------------BMI-----------")
weight = float(input("Enter your weight in kg:"))
height = float(input("Enter your height in m:"))

BMI = weight / (height**2)

if BMI < 18.5:
  print(f"Your BMI is {BMI} so you are underweight.")
elif BMI >=18.5 and BMI <=24.9:
  print(f"Your BMI is {BMI} so you are normal.")
elif BMI>=25 and BMI<=29.9:
  print(f"Your BMI is {BMI} so you are overweight.")
else:
  print(f"Your BMI is {BMI} so you are obese.")
```

```python
print("--------------name analyser----------")
name = input("Enter your full name:").strip()

title_name = name.title()
print("Title case of the name:", title_name)

initials = "".join([word[0].upper() + "." for word in name.split()])
print("Initials:", initials)

vowels = "AEIOUaeiou"
vowel_count = sum(1 for char in name if char in vowels)
print("the number of vowels:",vowel_count)
```
