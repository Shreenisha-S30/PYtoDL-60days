PRACTICE TASK

Write three programs: 1 a grade calculator for 5 subjects that also prints pass/fail; 2 the multiplication table for any number up to 20; 3 all primes between two numbers, and the count of primes.

```python
marks = []
for i in range(1, 6):
    score = float(input(f"Enter marks for subject {i} (out of 100): "))
    marks.append(score)
total = sum(marks)
percentage = total / 5
if any(mark < 36 for mark in marks):
    status = "Fail"
    grade = "F"
else:
    status = "Pass"
    if percentage >= 90:
        grade = "O"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    else:
        grade = "D"

print(f"\nTotal Marks: {total}/500")
print(f"Percentage: {percentage:.2f}%")
print(f"Result: {status} | Grade: {grade}")
```

```python
num = int(input("Enter the number to get multiplication table:"))
for i in range(1,21):
  print(f"{num} x {i} = {num*i}")
```

```python
start = int(input("Enter starting number:"))
end = int(input("Enter ending number:"))
count = 0
for num in range(start, end + 1):
  if num > 1:
    for i in range(2,num):
      if num % i == 0:
        break
    else:
      print(num, end= " , ")
      count = count+1
print("\ntotal primes:", count)
```
