# **PRACTICE** **TASK**

Take the marks of 10 students in a list and print the maximum, minimum, average, second-largest, and the marks above the average. Store each student as a tuple (name, marks) and print the topper's name.

```python
marks = []
students = []
for i in range(1, 11):
  name = input(f"Enter student {i} name: ")
  student_marks = int(input(f"Enter student {i} marks: "))
  marks.append(student_marks)
  students.append((name, student_marks))
print(marks)
max_marks = max(marks)
min_marks = min(marks)
avg_marks = sum(marks) / len(marks)
second_largest = 0
for m in marks:
  if m > second_largest and m < max_marks:
    second_largest = m
above_average = []
for m in marks:
  if m > avg_marks:
    above_average.append(m)
topper_name = ""
for name, m in students:
  if m == max_marks:
    topper_name = name
print("The maximum marks:", max_marks)
print("The minimum marks:", min_marks)
print("The average marks:", avg_marks)
print("The second largest:", second_largest)
print("Marks above average:", above_average)
print("Topper's name:", topper_name)
```
